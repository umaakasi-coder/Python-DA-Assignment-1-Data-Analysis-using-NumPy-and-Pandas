```python
#3(1). Array Operations:

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Convert to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

```python
#3(2).Find the maximum, minimum, and mean temperatures for the week.

max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)
print("Mean Temperature:", mean_temp)
```

```python
#4. Array Slicing and Indexing:

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

#Extract the temperatures for the first three days of the week.

first_three_days = temperatures_w1[:3]

print("First three days:", first_three_days)

#Extract the temperatures for the weekend (last two days).

weekend_temps = temperatures_w1[-2:]

print("Weekend temperatures:", weekend_temps)

#Extract the temperatures for the middle three days of the week.

middle_three_days = temperatures_w1[2:5]

print("Middle three days:", middle_three_days)
```

```python
#5. Create a 2D Array:

import numpy as np

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])

print(temperatures)
```

```python
#6(2)Extract the temperatures for each week
#Week1

week1 = temperatures[0]

print("Week 1 Temperatures:", week1)

#Week2

week2 = temperatures[1]

print("Week 2 Temperatures:", week2)
```

```python
#The weekends (last two days) for both weeks.

weekend_temps = temperatures[:, -2:]

print("Weekend Temperatures for Both Weeks:")
print(weekend_temps)
```

```python
#1. Creating Pandas Series:
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
#2. Indexing and Slicing:
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

# 1. First rank student (position-based)
print("1st Rank Mark:", marks.iloc[0])

# 2. Top 3 ranks using loc
print("\nTop 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# 3. Third rank student using iloc
print("\n3rd Rank Mark:", marks.iloc[2])

# 4. Marks greater than 90
print("\nMarks > 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Series:

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Original Series:")
print(marks)

#1. Modify the mark of the student with the 1st rank to 100

marks['Rank1'] = 100

print("After updating Rank1:")
print(marks)
```

```python
#2. Remove the entry corresponding to the last rank

marks = marks.drop('Rank5')

print("After removing Rank5:")
print(marks)
```

```python
#3. Compute the CGPA by dividing each mark by 10

cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
#Pandas DataFrame:
#1. Creating Pandas DataFrame:
# Create a Pandas DataFrame named transactions with the following data:

import pandas as pd

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
#2. Data Exploration:
#2(1).Display the transactions DataFrame and its basic information, including
#head, tail, shape, column names, and data types.
#2(1.a).Head (first 5 rows)

print(transactions.head())
```

```python
#2(1.b).Tail (last 5 rows)

print(transactions.tail())
```

```python
#2(1.c).Shape

print("Shape:", transactions.shape)
```

```python
#2(1.d).Column Names

print("Columns:", transactions.columns.tolist())
```

```python
#2(1.e).Data Types

print(transactions.dtypes)
```

```python
#Complete Information

transactions.info()
```

```python
#2(2).Display only the 'ProductCategory' and 'Amount' columns.

print(transactions[['ProductCategory', 'Amount']])
```

```python
#2(3).Retrieve the last 3 columns of the DataFrame using iloc or loc.

print(transactions.iloc[:, -3:])
```

```python
#2(4).Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)
```

```python
#2(5).Find the value counts for the 'ProductCategory' column.

print(transactions['ProductCategory'].value_counts())
```

```python
#2(6).Find the unique values in the 'Region' column.

print(transactions['Region'].unique())
```

```python
#2(7).Group by 'Region' and find the mean amount for each region.

mean_amount = transactions.groupby('Region')['Amount'].mean()

print(mean_amount)
```

```python
#3. Manipulating the DataFrame:
#1. Modify the Amount for TransactionID 102 to 165

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)
```

```python
#3(2).Add a new column 'Discount' by calculating 10% of the 'Amount'.

transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
#3(3). Remove the row with TransactionID 109

transactions = transactions[transactions['TransactionID'] != 109]

print(transactions)
```

```python
#3(4). Delete the 'Discount' column

transactions.drop('Discount', axis=1, inplace=True)

print(transactions)
```
